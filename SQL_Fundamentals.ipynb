```python

import sqlite3
conn = sqlite3.connect(":memory:")
cursor = conn.cursor()
```

```python
cursor.executescript("""
CREATE TABLE customers (
    customer_id INTEGER PRIMARY KEY,
    city TEXT
);

CREATE TABLE orders (
    order_id INTEGER PRIMARY KEY,
    customer_id INTEGER,
    product TEXT,
    quantity REAL,
    unit_price REAL,
    status TEXT
);

INSERT INTO customers VALUES
(1, 'Pune'),
(2, 'Mumbai'),
(3, 'Nashik');

INSERT INTO orders VALUES
(1001, 1, 'Laptop', 1, 50000, 'Completed'),
(1002, 2, 'Mouse', 2, 500, 'Completed'),
(1003, 1, 'Keyboard', 1, 1500, 'Completed'),
(1004, 3, 'Mouse', 3, 500, 'Completed'),
(1005, 2, 'Monitor', 1, 12000, 'Cancelled'),
(1006, 2, 'Keyboard', 2, 1500, 'Completed'),
(1007, 1, 'Laptop', 1, 48000, 'Completed'),
(1008, 3, 'Mouse', 1, 500, 'Completed'),
(1009, 1, 'Mouse', NULL, 500, 'Completed'),
(1010, 2, 'Monitor', 1, NULL, 'Completed');
""")
```

```python
cursor.execute("""
SELECT *
FROM orders;
""")
rows = cursor.fetchall()
for row in rows:
    print(row)
```

```python
cursor.execute("""
SELECT *
FROM orders
WHERE status = "Completed";
""")
rows = cursor.fetchall()
for row in rows:
    print(row)
```

```python
cursor.execute("""
SELECT *
FROM orders
WHERE unit_price > 1000;
""")
rows = cursor.fetchall()
for row in rows:
    print(row)
```

```python
cursor.execute("""
SELECT *
FROM orders
ORDER BY unit_price DESC
LIMIT 5;
""")
rows = cursor.fetchall()
for row in rows:
    print(row)
```

```python
#how many orders are in the database?
cursor.execute("""
SELECT COUNT(*)
FROM orders;
""")
rows = cursor.fetchall()
for row in rows:
    print(row)
```

```python
#how many completed orders are there?
cursor.execute("""
SELECT COUNT(*)
FROM orders
WHERE status = "Completed";
""")
rows = cursor.fetchall()
for row in rows:
    print(row)
```

```python
#what are the total revenue from valid completed orders?
cursor.execute("""
SELECT SUM(unit_price * quantity)
FROM orders
WHERE status = "Completed";
""")
rows = cursor.fetchall()
for row in rows:
    print(row)
```

```python
#what are the avarage order revenue?
cursor.execute("""
SELECT AVG(unit_price * quantity)
FROM orders
WHERE status = "Completed";
""")
rows = cursor.fetchall()
for row in rows:
    print(row)
```

```python
#which product generate the highest revenue?
cursor.execute("""
SELECT product, SUM(unit_price * quantity)
FROM orders
GROUP BY product
ORDER BY SUM(unit_price * quantity) DESC
LIMIT 1;
""")
rows = cursor.fetchall()
for row in rows:
    print(row)
```

```python
#which city generate the highest revenue?
cursor.execute("""
SELECT city, SUM(unit_price * quantity)
FROM orders
JOIN customers ON orders.customer_id = customers.customer_id
GROUP BY city
ORDER BY SUM(unit_price * quantity) DESC
LIMIT 1;
""")
rows = cursor.fetchall()
for row in rows:
    print(row)
```
