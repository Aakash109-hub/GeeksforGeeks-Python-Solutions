```python
import pandas as pd
import numpy as np
```

```python
data = {
    "order_id": [1001, 1002, 1003, 1004, 1005,
                 1006, 1007, 1008, 1009, 1010],
     "city": ["Pune", "Mumbai", "Pune", "Nashik",
             "Mumbai", "Mumbai", "Pune", "Nashik",
             "Pune", "Mumbai"],
    "product":["Laptop", "Mouse", "Keyboard", "Mouse",
                "Monitor", "Keyboard", "Laptop", "Mouse",
                "Mouse", "Monitor"],
    "quantity": [1, 2, 1, 3, 1, 2, 1, 1, np.nan, 1],
    "unit_price": [50000, 500, 1500, 500, 12000,
                   1500, 48000, 500, 500, np.nan],
     "status": ["Completed", "Completed", "Completed",
               "Completed", "Cancelled", "Completed",
               "Completed", "Completed", "Completed",
               "Completed"]
}
```

```python
df = pd.DataFrame(data)
```

```python
clean_df = df[df["status"] == "Completed"].copy()
```

```python
clean_df
```

```python
clean_df = clean_df.dropna(
    subset=["quantity", "unit_price"]
)
```

```python
clean_df["revenue"] = (
    clean_df["quantity"] * clean_df["unit_price"]
)
```

```python
clean_df
```

```python
#what is the total revenue?
total_revenue = clean_df["revenue"].sum()
total_revenue
```

```python
#Which city generate the highest revenue?
revenue_of_cities = clean_df.groupby("city")["revenue"].sum()
highest_city = revenue_of_cities.idxmax()
highest_amount = revenue_of_cities.max()
print(f"{highest_city}:{highest_amount}")
```

```python
#which product generate the highest revenue?
high_rev_prod = clean_df.groupby("product")["revenue"].sum()
high_prod = high_rev_prod.idxmax()
highest_amount = high_rev_prod.max()
print(f"{high_prod}:{highest_amount}")
```

```python
#What is the average revenue per order?
avg_rev_per_ord = clean_df["revenue"].mean()
avg_rev_per_ord
```

```python
#How many valid completed orders remain after cleaning?
valid_orders = clean_df[clean_df["status"] == "Completed"].shape[0]
valid_orders
```

```python
#Percentage contribution of the each city to total revenue
per_contri_city = (revenue_of_cities/total_revenue)*100
per_contri_city
```
